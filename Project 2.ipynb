## Data Exploration & Cleaning

```python
import pandas as pd
```

* Read the data from the `data.csv` file.
* Explore the data

```python
df = pd.read_csv('data.csv')
df
```

* Check for missing values.

```python
df.isnull().sum()
```

* Check for duplicate records.

```python
df.duplicated().sum()
```

* Check the data types of all columns.

```python
df.dtypes
```

* Convert columns to the appropriate data types where necessary:

  * Convert `Invoice Date` to datetime.
  * Convert numeric and currency columns to numeric values.

```python
df['Invoice Date'] = pd.to_datetime(df['Invoice Date'])
df['Price per Unit'] = (
        df['Price per Unit']
        .str.strip()
        .str.replace('$', '')
        .str.replace(',', '')
        .astype(float)
)
df['Units Sold'] = (
        df['Units Sold']
        .str.strip()
        .str.replace('$', '')
        .str.replace(',', '')
        .astype(float)
)
df['Total Sales'] = (
        df['Total Sales']
        .str.strip()
        .str.replace('$', '')
        .str.replace(',', '')
        .astype(float)
)
df['Operating Profit'] = (
        df['Operating Profit']
        .str.strip()
        .str.replace('$', '')
        .str.replace(',', '')
        .astype(float)
)
```

```python
print(df.dtypes)
df
```

* Create new columns from `Invoice Date`:

  * `day`
  * `month`
  * `year`

```python
df['day'] = df['Invoice Date'].dt.day
df['month'] = df['Invoice Date'].dt.month
df['year'] = df['Invoice Date'].dt.year
df
```

* Check the `Region` column and make sure there are no typos or inconsistent names.

```python
print(df.columns)
df.columns = df.columns.str.strip()
df.columns
```

* Drop unnecessary columns such as `Retailer ID`.

```python
df.drop('Retailer ID', axis=1, inplace=True)
```

* Handle missing values appropriately.

```python
df['Price per Unit'] = df['Price per Unit'].fillna(df['Total Sales'] / df['Units Sold'])
df['Region'] = df['Region'].fillna('Unknown')
```

### Sales Analysis

1. What is the total sales? 

```python
df['Total Sales'].sum()
```

2. What is the average sales for every region?

```python
df.groupby('Region')['Total Sales'].mean().round(2)
```

3. What is the average sales for every product?

```python
df['Product'] = df['Product'].str.replace('Men\'s aparel', 'Men\'s Apparel')
```

```python
df.groupby('Product')['Total Sales'].mean().round(2)
```

4. What is the count of each product?

```python
df['Product'].value_counts()
```

5. What is the top product by total sales?

```python
product_sales = (
    df.groupby('Product')['Total Sales']
    .sum().round(2).sort_values(ascending=False)
)
print(product_sales)
product_sales.index[0]
```

6. What is the lowest-performing product by total sales?

```python
product_sales.index[-1]
```

### Time Analysis

7. How many products were sold each year?

```python
df['year'].value_counts()
```

8. How have total sales changed over the years?

```python
df.groupby('year')['Total Sales'].sum().round(2)
```

9. How do sales vary month over month?

```python
df.groupby('month')['Total Sales'].sum().round(2)
```

### Retailer Analysis

10. Which retailer has the highest total sales?

```python
retailer_sales = (
    df.groupby('Retailer')['Total Sales']
    .sum().round(2).sort_values(ascending=False)
)
print(retailer_sales)
retailer_sales.index[0]
```

11. Which retailer has the highest operating profit?

```python
retailer_price = (
    df.groupby('Retailer')['Operating Profit']
    .sum().round(2).sort_values(ascending=False)
)
print(retailer_price)
retailer_price.index[0]
```

### Product & Profit Analysis

12. What is the most profitable product?

```python
product_profit = (
    df.groupby('Product')['Price per Unit']
    .sum().round(2).sort_values(ascending=False)
)
print(product_profit)
product_profit.index[0]
```

13. What is the average price per unit for each product?

```python
df.groupby('Product')['Price per Unit'].mean().round(2).sort_values(ascending=False)
```

### Geographic Analysis

14. Which region has the highest total sales?

```python
region_sales = (
    df.groupby('Region')['Total Sales']
    .sum().round(2).sort_values(ascending=False)
)
print(region_sales)
region_sales.index[0]
```

15. Which city contributes the most to total sales?

```python
city_sales = (
    df.groupby('city')['Total Sales']
    .sum().round(2).sort_values(ascending=False)
)
print(city_sales)
city_sales.index[0]
```
